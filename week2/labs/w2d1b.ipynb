## 4. Logic Errors

Not every bug causes an error message.

A **logic error** happens when the code runs, but gives the wrong answer.

These can be harder because Python does not always tell us what went wrong.

### 4.1

The function below should return the numbers from `1` to `n`, including `n`.

When you run the function below, you will see that it currently misses one number (known as an off-by-one error). Fix it.

````python
def numbers_from_one_to_n(n):
    nums = []
    for i in range(1, n):
        nums.append(i)
    return nums

assert_equal(got=numbers_from_one_to_n(5), want=[1, 2, 3, 4, 5])
assert_equal(got=numbers_from_one_to_n(1), want=[1])
````

### 4.2

The function below should multiply all the numbers in a list.

When you run the function below, you will see that it currently gives the wrong answer. Fix it.

````python
def product(nums):
    total = 0
    for num in nums:
        total *= num
    return total

assert_equal(got=product([2, 3, 4]), want=24)
assert_equal(got=product([5]), want=5)
````

### 4.3

The function below should count how many even numbers are in a list.

When you run the function below, you will see that it currently gives the wrong answer. Fix it.

````python
def count_evens(nums):
    count = 0
    for num in nums:
        if num % 2 == 0:
            num += 1
    return count

assert_equal(got=count_evens([1, 2, 3, 4, 5, 6]), want=3)
assert_equal(got=count_evens([1, 3, 5]), want=0)
````

## 5. Debugging Strategy

When your code breaks, use this checklist:

1. Read the last line of the error message.
2. Find the error type.
3. Find the line where Python got stuck.
4. Print important variables before that line.
5. Try a smaller example.
6. Fix one thing at a time.

In this section, you will debug functions that have more than one problem.

### 5.1 Add Print Statements

The function below is supposed to return the largest number in a list.

It has two bugs. Add print statements if they help you understand what is happening, then fix the function.

````python
def find_max(nums):
    biggest = 0
    for num in nums:
        if num > biggest:
            biggest = nums
    return biggest

assert_equal(got=find_max([1, 5, 2]), want=5)
assert_equal(got=find_max([-5, -2, -10]), want=-2)
````

### 5.2 Final Debugging Practice

The function below should return a list containing only the words with length at least `min_length`.

Fix the code so all tests pass. Hint: there may be more than one bug!

````python
def long_words(words, min_length):
    result = []
    for i in range(len(words) + 1):
        word = words[i]
        if len(word) > min_length:
            result.append(words)
    return result

assert_equal(got=long_words(["hi", "banana", "cat", "python"], 4), want=["banana", "python"])
assert_equal(got=long_words(["a", "to", "sun"], 2), want=["to", "sun"])
````

# Nested Loops Recap
We can actually have nested ```for``` loops. This means we can place a ```for``` loop inside another ```for``` loop. These are used when we need to do something repeatedly inside another task that's also being repeated.

<b>Example 1:</b>
```python
for i in range(3):
    print("I'm the outer loop! Currently, i is", i)
    for j in range(4):
        print("Inner loop! j is", j)
```

```for i in range(3):``` is considered the outer loop. The loop nested inside is called the ```inner``` loop. In this example, the outer loop will repeat ```3``` times, and with each iteration, it will execute the inner loop ```4``` times.

First, run the code block below:

````python
for i in range(3):
    print("I'm the outer loop! Currently, i is", i)
    for j in range(4):
        print("Inner loop! j is", j)
````

Let's step through the code to see what exactly is happening.

First, ```i``` takes on the value ```0```, it will execute the inner loop ```4``` times. So, ```j``` takes on the values ```0```, ```1```, ```2```, ```3```, and prints ```"Inner Loop! j is, j"``` for each value.  

Next, ```i``` takes on the value ```1```, it will execute the inner loop ```4``` times. So, ```j``` takes on the values ```0```, ```1```, ```2```, ```3```, and prints ```"Inner Loop! j is, j"``` for each value.  

Lastly, ```i``` takes on the value ```2```, it will execute the inner loop ```4``` times. So, ```j``` takes on the values ```0```, ```1```, ```2```, ```3```, and prints ```"Inner Loop! j is, j"``` for each value.  


<b>Example 2</b>

We can also use nested ```for``` loops with lists!
```python
students = ["Mary", "Zara", "Edward", "Alex"]
school_materials = ["textbook", "lab equipment", "notebooks"]

for student in students:
    for material in school_materials:
        print(student, "received", material)
```

Here, ```for student in students:``` is the outer loop and the inner loop is ```for material in school_materials:```. We are assigning every ```student``` with the necessary school materials.

Run the code and examine the output:

````python
students = ["Mary", "Zara", "Edward", "Alex"]
school_materials = ["textbook", "lab equipment", "notebooks"]

for student in students:
    for material in school_materials:
        print(student, "received", material)
````

Let's step through the code to see what exactly is happening.

First, ```student``` takes on the value ```"Mary"```. It will execute the inner loop, which steps through each element in ```school_materials```: ```material``` will be assigned ```"textbook"```, then ```"lab equipment"```, then ```"notebooks"```.  

Second, ```student``` takes on the value ```"Zara"```. It will execute the inner loop, which steps through each element in ```school_materials```: ```material``` will be assigned ```"textbook"```, then ```"lab equipment"```, then ```"notebooks"```.  

Third, ```student``` takes on the value ```"Edward"```. It will execute the inner loop, which steps through each element in ```school_materials```: ```material``` will be assigned ```"textbook"```, then ```"lab equipment"```, then ```"notebooks"```.  

Lastly, ```student``` takes on the value ```"Alex"```. It will execute the inner loop, which steps through each element in ```school_materials```: ```material``` will be assigned ```"textbook"```, then ```"lab equipment"```, then ```"notebooks"```.
