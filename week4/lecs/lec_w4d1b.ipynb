## Skittle-up 📈

Let's make a cup with hundredsssss of Skittles...

```python
# Make a cup with some red, purple, and yellow skittles
cup = []
for i in range(40):
    cup.append("red")
for i in range(100):
    cup.append("purple")
for i in range(60):
    cup.append("yellow")
```

```python
print(cup[2])
```

```python
print("Our cup has", len(cup), "Skittles")
```

```python
print(cup[:20])   # just the first 20
print(len(cup))
```

### Running the experiment

Just like with our physical cups, we'll:
- Sample many times from our digital cup
- Try different sample sizes to see how our estimates improve

```python
from jamcoders.random import visualize
```

The `visualize` function will show us:
- What's in our cup 🥤 (True distribution)
- What we saw when sampling (Sampled distribution)

The sampled distributions get closer to the true distribution as we take more samples

```python
sample_sizes = [15, 150, 1000, 5000]
visualize(cup, sample_sizes)
```

Can we make a smaller model that captures sampling from `big_cup`?
- HINT: Do we need to remember each Skittle, or are the **Probabilities** (ratios) enough?

```python
# Input: Cup (list of strings)
# Output: Dictionary mapping string to its probability (ratio)
def compute_probs(cup):
    counts = {}
    for skittle in cup:
        if skittle not in counts:
            counts[skittle] = 0
        counts[skittle] += 1
    probs = {}
    for skittle in counts:
        probs[skittle] = counts[skittle] / len(cup)
    return probs

print(compute_probs(big_cup))
```
