## Functions

__Function__

Machine taking inputs to output

Define __once__ use __many times__

`print` is not `return`!!!

````python
def f(a,b):
    return a+b
````

````python
# can call many times
print(f(5,7))
print(f(2,4))
````

````python
# can even compose calls
print( f( f(1,2) , f(3,4) ) )
````

Function is a __black box__ 

````python
def f1(x):
    return x+x

def f2(x):
    a = x
    b = x
    return a+b

def f3(y):
    return 2*y

def f4(x):
    res = 0
    for i in range(x):
        res += 2
    return res
````

````python
for x in range(10):
    if f1(x)==f2(x) and f2(x)==f3(x) and f3(x)==f4(x):
        print(f"They all agree on {x}")
    else:
        print(f"Disagreement on {x}!!")
````

# Loops

__English:__ Do `blah` 10 times

__Python:__ 
```python
for i in range(10):
    blah
```

````python
for something_completely_different in range(10):
    print(something_completely_different,"*")
````

__English:__ Go over every element `x` of a list `L` and do `blah` with `x`

__Python:__
```python
for x in L:
    blah
```

````python
M = ["I", "will", "miss", "Jamcoders" ]
for x in range(len(M)):
    print(M[x])
````

**While loop:**

**English:** Do `blah` as long as `condition` is True

**Python:**

```python
while cond:
    blah
```

````python
keep_going = True
while keep_going:
    answer = input("Should I keep going? Answer 'q' to quit.")
    if answer=="q":
        keep_going = False
print("done")
````

# Lists and Strings

List of elements, can contain elements of any type: `int`, `float`, `str` and even other lists!

* `L[i]`: the element that is `i+1`-th from beginning in the list `L` 
* `L[-i]`: element that is `i`-th from end in the list `L`
* `L[i:j]`: elements between positions `i` to `j-1`.
* `L + M`: concatenate two lists

Strings:
* `s[i]`: character in `i+1`-th position.
* `s[-i]`: character that is `i`-th from end in the list `s`
* `s[i:j]`: characters between positions `i` to `j-1`.
* `s + t`: concatenate two strings

`L` and `M` for lists and `s` and `t` for strings is just **convention**

````python
from termcolor import colored
rows = [
    "..................................................",
    "###.#.#...#......#...#.###..##..##....#.#..#..#.#.",
    ".#..#.#...#......##.##..#..#...#......#.#.#.#.#.#.",
    ".#....#...#......#.#.#..#...#...#......#..#.#.#.#.",
    ".#....#...#......#...#..#....#...#.....#..#.#.#.#.",
    "###...###.###....#...#.###.##..##......#...#..###.",
    "..................................................",
]
for r in range(len(rows)):
    line = ""
    for c in range(len(rows[r])):
        if rows[r][c] == "#":
            bg = "on_yellow"
        elif (r + c) // 3 % 2 == 1:
            bg = "on_green"
        else:
            bg = "on_black"
        line += colored("  ", "black", bg)
    print(line)
````
