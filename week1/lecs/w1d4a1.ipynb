````python
def did_player_win(player,board):
    # horizontal
    if (board[0][0]==player) and (board[0][1]==player) and (board[0][2]==player):
        return True
    if (board[1][0]==player) and (board[1][1]==player) and (board[1][2]==player):
        return True
    if (board[2][0]==player) and (board[2][1]==player) and (board[2][2]==player):
        return True

    # vertical
    if (board[0][0]==player) and (board[1][0]==player) and (board[2][0]==player):
        return True
    if (board[0][1]==player) and (board[1][1]==player) and (board[2][1]==player):
        return True
    if (board[0][2]==player) and (board[1][2]==player) and (board[2][2]==player):
        return True

    # diagonal
    if (board[0][0]==player) and (board[1][1]==player) and (board[2][2]==player):
        return True

    if (board[2][0]==player) and (board[1][1]==player) and (board[0][2]==player):
        return True

    return False
````

````python
board = [[".",".","."], [".",".","."],[".",".","."]]
player = "X"
for i in range(9):
    clear_output()
    print_nicer(board)
    print(f"Player {player}'s turn")
    i = int(input("i:"))
    j = int(input("j:"))
    board[i][j] = player
    if did_player_win(player,board):
        clear_output()
        print_nicer(board)
        print(f"Player {player} won!")
        break
    if player == "X":
        player = "O"
    else:
        player = "X"
````

Can we make `did_player_win` simpler using loops?

````python
def did_player_win(player,board):
    # horizontal
    for i in range(3):
        if (board[i][0]==player) and (board[i][1]==player) and (board[i][2]==player):
            return True

    # vertical
    for j in range(3):
        if (board[0][j]==player) and (board[1][j]==player) and (board[2][j]==player):
            return True

    # diagonal
    if (board[0][0]==player) and (board[1][1]==player) and (board[2][2]==player):
        return True

    if (board[2][0]==player) and (board[1][1]==player) and (board[0][2]==player):
        return True

    return False
````

# Advanced for loop keywords

````python
# This is a reminder of what a loop looks like

for i in range(20):
    print('still in loop')
print('Loop completed!')
````

````python
# There are two keywords for loops that you will need to know
# The first is "break"
# Break will exit a for loop once it is executed

for i in range(4):
    print(i)
    if i == 2:
        break
    print("---")
print('Loop completed!')
````

````python
# The other keyword you need to know is "continue"
# "continue" will make it so that the rest of the loop body is not executed
# but ONLY for the time it is called
# Notice that "10" is not printed


for i in range(4):
    print(i)
    if i == 2:
        continue
    print("---")
print('Loop completed!')
````

````python
# You can have continue and break in the same loop

for i in range(6):
    if i == 2:
        continue
    if i == 4:
        break
    print(i)
print('Loop completed!')
````

# While loops

````python
# This is the generic syntax for a while loop

while CONTIDION:
    LOOP_BODY
````

````python
# A "while loop" is a loop that executes while CONDITION is true
# The CONDITION is evaluated once the loop body finishes, and only then

i = 0
while i < 10:
    print(i)
    i += 1
print('Loop completed!')
````

````python
# If CONDITION is False at the start of the loop, the loop body will not execute!

i = 0
while i < 0:
    print(i)
    i += 1
print('Loop completed!')
````

````python
# What is different about this and the one two above?

i = 0
while i < 10:
    i += 1
    print(i)
print('Loop completed!')
````

````python
# WARNING!!!!
# You can have loops that run forever.
# These are really really bad.
# Make sure the loop terminates.

while True:
    print('hi!')
````

````python
# WARNING!!!!!
# Even if you don't mean to, the loop can run forever.
# DO NOT do this.

i = 0
while i < 10:
    print(i)
print('Loop completed!') # This will never execute!
````

````python
# You can use "continue" in while loops
# The syntax and behavior is exactly the same as in for loop

i = 0
while i < 10:
    i += 1
    if i == 5:
        continue
    print(i)
print('Loop completed!')
````

````python
# WARNING!!!!
# This loop will also run forever. Why?
# Don't do this.

i = 0
while i < 10:
    if i == 5:
        continue
    print(i)
    i += 1
print('Loop completed!')
````

````python
# You can also use "break" in while loops
# The syntax and usage is exactly the same as in for loops

i = 0
while True:
    if i > 10:
        break
    print(i)
    i += 1
````

# Chat


````python
from IPython.display import Markdown, display
from openai import OpenAI
import os


def chat(prompt):
    """Call GPT-5.5 with high reasoning effort and return the answer as a string."""
    if not os.environ.get("OPENAI_API_KEY"):
        raise RuntimeError("Set OPENAI_API_KEY in the environment before calling chat().")

    status = display(
        Markdown("Working: calling `gpt-5.5` with high reasoning effort..."),
        display_id=True,
    )
    if status is None:
        print("Working: calling gpt-5.5 with high reasoning effort...")

    def update_status(message):
        if status is None:
            print(message)
        else:
            status.update(Markdown(message))

    try:
        client = OpenAI()
        response = client.responses.create(
            model="gpt-5.5",
            reasoning={"effort": "high"},
            input=[{"role": "user", "content": prompt}],
        )
        answer = response.output_text or ""
        update_status(f"Done. Returned {len(answer)} characters.")
        return str(answer)
    except Exception as exc:
        update_status(f"OpenAI call failed: `{type(exc).__name__}`")
        raise
````

````python
answer = chat("In one short sentence, what does a Python loop do?")
print(answer)
````

````python
def translate_to_patois(sentence):
    msg = f"Translate the following sentence to Jamaican Patois. Just return the sentence.\n{sentence}"
    return chat(msg)

print(translate_to_patois("I am at the back of the class and I don't have any subject."))
````
