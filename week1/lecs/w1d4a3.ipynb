````python
def print_nicer_row(row):
    print(row[0]+"┃"+row[1]+"┃"+row[2])
    
def print_nicer(board):
    clear_output()
    separator = "━╋━╋━"
    print_nicer_row(board[0])
    print(separator)
    print_nicer_row(board[1])
    print(separator)
    print_nicer_row(board[2])
                
print_nicer(board)
````

````python
def did_player_win(player,board):   
    # horizontal
    if (board[0][0]==player) and (board[0][1]==player) and (board[0][2]==player):
        return True
    if (board[1][0]==player) and (board[1][1]==player) and (board[1][2]==player):
        return True
    if (board[2][0]==player) and (board[2][1]==player) and (board[2][2]==player):
        return True
    
    # vertical    
    if (board[0][0]==player) and (board[1][0]==player) and (board[2][0]==player):
        return True
    if (board[0][1]==player) and (board[1][1]==player) and (board[2][1]==player):
        return True
    if (board[0][2]==player) and (board[1][2]==player) and (board[2][2]==player):
        return True
    
    # diagonal    
    if (board[0][0]==player) and (board[1][1]==player) and (board[2][2]==player):
        return True
    
    if (board[2][0]==player) and (board[1][1]==player) and (board[0][2]==player):
        return True
    
    return False
````

````python
from IPython.display import Markdown, display
from openai import OpenAI
import os


def chat(prompt):
    """Call GPT-5.5 with high reasoning effort and return the answer as a string."""
    if not os.environ.get("OPENAI_API_KEY"):
        raise RuntimeError("Set OPENAI_API_KEY in the environment before calling chat().")

    status = display(
        Markdown("Working: calling `gpt-5.5` with medium reasoning effort..."),
        display_id=True,
    )
    if status is None:
        print("Working: calling gpt-5.5 with medium reasoning effort...")

    def update_status(message):
        if status is None:
            print(message)
        else:
            status.update(Markdown(message))

    try:
        client = OpenAI()
        response = client.responses.create(
            model="gpt-5.5",
            reasoning={"effort": "medium"},
            input=[{"role": "user", "content": prompt}],
        )
        answer = response.output_text or ""
        update_status(f"Done. Returned {len(answer)} characters.")
        return str(answer)
    except Exception as exc:
        update_status(f"OpenAI call failed: `{type(exc).__name__}`")
        raise
answer = chat("In one short sentence, what does a Python loop do?")
print(answer)
````

# Tic Tac Toe

````python
def make_move(player,board):
    print(f"Player {player}'s turn")
    i = int(input("i:"))
    j = int(input("j:"))
    if board[i][j] != '.': # check valid
        print(f"ERRROR: {i},{j} is taken!")
        return
    board[i][j] = player
````

````python
def tic_tac_toe():
    board = [[".",".","."], [".",".","."],[".",".","."]]
    player = "X"
    for turn in range(9):        
        print_nicer(board)
        make_move(player,board)
        if did_player_win(player,board):
            print_nicer(board)
            print(f"Player {player} won! 🎉🎉🎉")   
            return
        # flip players
        if player == "X": 
            player = "O"
        else:
            player = "X"
    
    # loop ended: tie
    print_nicer(board)
    print("Tie!")  
````

````python
tic_tac_toe()
````

Write a function

```python
def board_to_string(board):
    ...
    # returns a string that includes three lines with just the symbols in the board
```

**Example:** If `board = [['X', 'O', 'X'], ['X', 'O', 'O'], ['O', 'X', 'X']]`
then `board_to_string(board)` will be

```
XOX
XOO
OXX
```


````python
def board_to_string(board):
    result = ""
    for row in board:
        for cell in row:
            result += cell
        result += "\n"
    return result

board = [['X', 'O', 'X'], ['X', 'O', 'O'], ['O', 'X', 'X']]
x = board_to_string(board)
print(x)
        
````

## Slicing

If `L` is a list then `L[2:7]` is the elements of `L` in positions 2 until 6. Same for string.

````python
L = ["a","b","c","d","e","f","g"]
# What will be printed
print(L[2:4])
print(L[4:7])
````

If `L` is a list then `L[2:]` is the elements of `L` in positions 2 until the end, `L[:7]` is the elements of `L` from the beginning until position 6. Same for strings.

````python
s = "JamCoders"
a = s[:3]
b = s[3:]
print(a)
print(b)
````

````python
def mystery(s):
    result = ""
    for i in range(len(s)):
        if (s[i] == "o"):
            result += "*"
        else:
            result += s[i]
        print(result)
    return result

# What will be printed?
print(mystery("JamCoders"))
````

````python
def find_comma(s):
    for i in range(len(s)):
        if s[i]==',':
            return i
    return -1
print(find_comma("12,4"))
````

````python
def find_two_commas(s):
    # returns a list [first,second] with the first and second
    # commas in the string (or -1 for missing commas)
    first = -1
    for i in range(len(s)):
        if s[i]==',':
            if first==-1:
                first=i
            else:
                return [first,i]
    return [first,-1]
````
