If $x$ is a number, then the _integer square root_ of $x$ is the largest integer $y$ such that $y^2 \leq x$.

* The integer square root of 9 is 3
* The integer square root of 10 is 3
* The integer square root of 17 is 4

Write a function `isqrt(x)` that on input a positive integer `x`, outputs the integer square root of `x`

__Hints:__ 

* The integer square root of $x$ is a number in $[0,1,2,\ldots,x]$

* It is _smallest_ $y$ such that $(y+1)^2 > x$.

````python
def isqrt(x):
    for i in range(x+1):
        if i*i==x:
            return i
        else:
            if i*i > x:
                return i-1

for x in [1,2,3,4,5]:
    print(f"isqrt({x}) equals {isqrt(x)}")
````

````python
def isqrt(x):
    for y in range(x+1):
        if (y+1)**2 > x:
            return y
    
````

````python
print(isqrt(25))
````

# Lists

List of elements, can contain elements of any type: `int`, `float`, `str` and even other lists!

* `L[i]`: the element that is `i+1`-th from beginning in the list `L` 
* `L[-i]`: element that is `i`-th from end in the list `L`
* `L[i:j]`: elements between positions `i` to `j-1`.
* `L.append(x)` adds `x` to the list `L`


__Question:__ Write function `prod` that on input a list `L` of numbers, outputs the product of the numbers

__Examples:__ `prod([1,2,3]) == 6` , `prod([10,2])==20`

````python
def prod(L):
    product = 1
    for i in L:
        product *= i
    return product

print(prod([1,2,3]))
````

# Advanced for loop keywords

````python
# This is a reminder of what a loop looks like

for i in range(20):
    print('still in loop')
print('Loop completed!')
````

````python
# There are two keywords for loops that you will need to know
# The first is "break"
# Break will exit a for loop once it is executed

for i in range(4):
    print(i)
    if i == 2:
        break
    print("---")
print('Loop completed!')
````

````python
# The other keyword you need to know is "continue"
# "continue" will make it so that the rest of the loop body is not executed
# but ONLY for the time it is called
# Notice that "10" is not printed


for i in range(4):
    print(i)
    if i == 2:
        continue
    print("---")
print('Loop completed!')
````

````python
# You can have continue and break in the same loop

for i in range(6):
    if i == 2:
        continue
    if i == 4:
        break
    print(i)
print('Loop completed!')
````

# While loops

````python
# This is the generic syntax for a while loop

while CONTIDION:
    LOOP_BODY
````

````python
# A "while loop" is a loop that executes while CONDITION is true
# The CONDITION is evaluated once the loop body finishes, and only then

i = 0
while i < 10:
    print(i)
    i += 1
print('Loop completed!')
````

````python
# If CONDITION is False at the start of the loop, the loop body will not execute!

i = 0
while i < 0:
    print(i)
    i += 1
print('Loop completed!')
````

````python
# What is different about this and the one two above?

i = 0
while i < 10:
    i += 1
    print(i)
print('Loop completed!')
````

````python
# WARNING!!!!
# You can have loops that run forever.
# These are really really bad.
# Make sure the loop terminates.

while True:
    print('hi!')
````

````python
# WARNING!!!!!
# Even if you don't mean to, the loop can run forever.
# DO NOT do this.

i = 0
while i < 10:
    print(i)
print('Loop completed!') # This will never execute!
````

````python
# You can use "continue" in while loops
# The syntax and behavior is exactly the same as in for loop

i = 0
while i < 10:
    i += 1
    if i == 5:
        continue
    print(i)
print('Loop completed!')
````

````python
# WARNING!!!!
# This loop will also run forever. Why?
# Don't do this.

i = 0
while i < 10:
    if i == 5:
        continue
    print(i)
    i += 1
print('Loop completed!')
````

````python
# You can also use "break" in while loops
# The syntax and usage is exactly the same as in for loops

i = 0
while True:
    if i > 10:
        break
    print(i)
    i += 1
````
