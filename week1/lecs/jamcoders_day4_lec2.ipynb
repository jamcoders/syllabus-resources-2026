````python
def enc(message):
    # take a string message
    # return version with a↦b, b↦c,...,z↦a
    result = ""
    lower = lower_case()
    for i in range(len(message)):
        letter = message[i]
        position = lower.find(letter)
        if position == -1:
            result += letter
            continue # skip the rest of the loop
        if position == 25:
            new_position = 0
        else:
            new_position = position + 1
        new_letter = lower[new_position]
        result += new_letter
    return result
print(enc("jam coders!!!AAA"))
````

````python
def dec(s):
    lower_case_letters = lower_case()
    res =""
    for a in s:
        i = lower_case_letters.find(a)
        res += lower_case_letters[(i-1) % 26]
    return res

encoded = enc("jamcoders")
decoded = dec(encoded)
print("Encoded", encoded, "Decoded", decoded)
````

````python
# A "while loop" is a loop that executes while CONDITION is true
# The CONDITION is evaluated once the loop body finishes, and only then

i = 0
while i < 10:
    print(i)
    i += 1
print('Loop completed!')
````

A _Palindrome_ is a string that is the same if you read it forward and backwards:

Which one is a Palindrome?

* `kayak`?

* `banana`?

* `racecar`?

* `aaabaaa`?

Write a function `pal4(s)` that takes as input `s` of length 4, and returns `True` if `s` is a Palindrome and `False` otherwise.  

````python
def pal4(s):
    if (s[0]==s[-1]) and (s[1]==s[-2]):
        return True
    return False

print(pal4("abba"))
print(pal4("abca"))
````

````python
print(pal4("abba"))
print(pal4("boaz"))
````

Write a function `pal6(s)` that takes as input `s` of length 6, and returns `True` if `s` is a Palindrome and `False` otherwise.  Use `or`

````python
def pal6(s):
    if s[0]!=s[-1] or s[1]!=s[-2] or s[2] != s[-3]:
        return False
    return True
````

````python
print(pal6("abccba"))
````

Write function `pal(s)` that works for Palindromes for every even length. As bonus, make it work for every length

````python
def pal(s):
    for i in range(len(s)//2):
        if s[i] != s[-i-1]:
            return False
    return True

print(pal("racecar"))
````
