# For loops

The format of a `for` loop is
```python
for VARIABLE in COLLECTION:
    FOR_LOOP_BODY
```

`COLLECTION` can be __list__, __range__, and other objects too  

````python
for x in [2,19,4]:
    y = 2*x
    print(y)
    
````

````python
# What will be printed?
result = 0
for x in [2,19,4]:
    print(f"Looping with x= {x}, result = {result}")
    print(f"  Updating result to {result+x}")
    result = result + x
print(result)
````

````python
for x in range(17):
    print(x)
````

````python
# What will be printed?
result = 0
for x in range(101): 
    result = result + x
    
print(result)
````

## Loops on lists

````python
# Lists can contain any type
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
````

````python
# You can put control flow in a loop
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(f"Looping with x={x}")
    if x == 'apple':
        print(x)
    else:
        print("(not printing anything)")
````

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x != 'apple':
        print(x)
````

Write a function `starts_with_b` that takes as input a list `L` of strings and returns the first element in the list that starts with the letter `b`. Otherwise it returns `None`.

```python
def starts_with_b(L):
   for elem in L:
      if elem[0]=="b":
        # do something
    # do more things
```

`print(starts_with_b(["apple", "banana", "cherry"])`

results in `banana`

````python
def starts_with_b(L):
    """Returns the first element in the list L 
    that starts with the letter b"""
    for elem in L:  
        if elem[0]=='b': 
            return elem
    return None
````

````python
print(starts_with_b(["apple","cherry","pear"]))
````

````python
# Let's see what's going on
def starts_with_b_print(L):
    """Returns the first element in the list L that starts with the letter b"""
    print(f"We received {L} with length {len(L)}")
    for elem in L:
        print(f"  looping on {elem}")
        if elem[0]=='b':
            print(f"    first letter of {elem} is `b`: returning {elem}")
            return elem
    print("The loop ended - returning None")
    return None
````

````python
starts_with_b_print(["apple","pear","cherry"])
````

````python
print( starts_with_b_print(["apple", "pear", "cherry"]) )
````

When you write a function, it makes some __assumptions__ on its input. If the assumptions are wrong then it may result in an error or _Exception_

````python
print(starts_with_b_print(["apple","boaz","banana"]))
````

````python
L = ["Apple","","banana"]
starts_with_b_print(L)
````

Sometimes because of errors, the function will not have an exception, but will return the wrong output - **this is worse!**

````python
print(starts_with_b_print("dabble"))
````

Always make sure to call a function with inputs that match its assumptions!

````python
# What will happen?
starts_with_b_print([])
````

## Options for range 

````python
# range with two arguments (call them n and m) is like looping over the list from n to m-1
for x in range(3, 6):
    print(x)
````

````python
# What will this do?
for x in range(3, 3):
    print(x)
````

````python
# Range with three arguments (call them n, m, and step)
# is like looping over the list from n to m-1, but incrementing by step
for x in range(3, 8, 3):
    print(x)
````

````python
# You can put loops inside loops
# ALWAYS make sure the variables for the loops are different
for i in range(3):
    print(f"Outer loop with i={i}")
    for j in range(4): # DO NOT USE i HERE
        print(f"  Inner loop with", "i=",i, "j=",j)
````

````python
def print_board(board):
    for row in range(8):
        for col in range(8):
            is_light_square = (row + col) % 2 == 0
            piece = board[row][col]

            if is_light_square:
                background = "on_light_grey"
                foreground = "black"
            else:
                background = "on_dark_grey"
                foreground = "white"

            print(colored(f" {piece} ", foreground, background), end="")
        print()
````
